```python
dicionario = "qwertyuiopasdfghjklzxcvbnm"
alfabeto = "abcdefghijklmnopqrstuvwxyz"
senha = "marcelo"
senha_criptografada = ""

for letra in senha:
    posicao = ord(letra) - ord("a") # 0 até 26
    senha_criptografada += dicionario[posicao]

print(senha_criptografada)
```
